# Multi-Class Overall Best Setup

This notebook reads the combined comparison results across all 5 classifiers (CART, Random Forest, KNN, SVM, XGBoost) and identifies the optimal intersection of season and cloud cover threshold based on average accuracy.

```python
import pandas as pd

# Read the combined sweep results
df = pd.read_csv('combined_comparison_results.csv')

# Calculate average accuracy and kappa for each (season, cloud cover) combination
intersection_scores = df.groupby(['season', 'cloud_cover'])[['accuracy', 'kappa']].mean().reset_index()
intersection_scores = intersection_scores.rename(columns={'accuracy': 'avg_accuracy', 'kappa': 'avg_kappa'})

# Find the combination with the highest average accuracy
best_intersection = intersection_scores.sort_values(by=['avg_accuracy', 'avg_kappa'], ascending=[False, False]).iloc[0]

print("="*60)
print("  MULTI-CLASS OVERALL BEST SETUP (ACROSS ALL 5 CLASSIFIERS)  ")
print("="*60)
print(f"Best Season:      {best_intersection['season']}")
print(f"Best Cloud Cover: {best_intersection['cloud_cover']}%")
print(f"Average Accuracy: {best_intersection['avg_accuracy']:.4f}")
print(f"Average Kappa:    {best_intersection['avg_kappa']:.4f}\n")

print("Breakdown for this best setup:")
best_setup_df = df[(df['season'] == best_intersection['season']) & (df['cloud_cover'] == best_intersection['cloud_cover'])]
for _, row in best_setup_df.iterrows():
    print(f"  {row['classifier']:20s} Accuracy: {row['accuracy']:.4f}   Kappa: {row['kappa']:.4f}")
```
